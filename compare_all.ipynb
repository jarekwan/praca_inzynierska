```python
%%writefile /content/drive/MyDrive/ml_project/compare_all.py
# -*- coding: utf-8 -*-
import os
import numpy as np
import pandas as pd

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

target_dir = "/content/drive/MyDrive/ml_project"

y_test_path = os.path.join(target_dir, "y_test.pkl")
y_pred_ml_path = os.path.join(target_dir, "y_pred_ml.pkl")
y_pred_tr_path = os.path.join(target_dir, "y_pred_traditional.pkl")

output_table_path = os.path.join(target_dir, "comparison_table.pkl")


# ---------------------------------------------------------
# compare_all
# ---------------------------------------------------------
def compare_all():

    # check files
    if not os.path.exists(y_test_path):
        raise FileNotFoundError("y_test.pkl not found")

    if not os.path.exists(y_pred_ml_path):
        raise FileNotFoundError("y_pred_ml.pkl not found")

    if not os.path.exists(y_pred_tr_path):
        raise FileNotFoundError("y_pred_traditional.pkl not found")

    # load data
    y_test = pd.read_pickle(y_test_path)
    y_pred_ml = pd.read_pickle(y_pred_ml_path)
    y_pred_tr = pd.read_pickle(y_pred_tr_path)

    # convert to numpy
    y_test = np.array(y_test)
    y_pred_ml = np.array(y_pred_ml)
    y_pred_tr = np.array(y_pred_tr)

    # ml metrics
    mae_ml = mean_absolute_error(y_test, y_pred_ml)
    rmse_ml = mean_squared_error(y_test, y_pred_ml, squared=False)
    mape_ml = np.mean(np.abs((y_test - y_pred_ml) / np.where(y_test == 0, 1, y_test))) * 100
    r2_ml = r2_score(y_test, y_pred_ml)

    # traditional metrics
    mae_tr = mean_absolute_error(y_test, y_pred_tr)
    rmse_tr = mean_squared_error(y_test, y_pred_tr, squared=False)
    mape_tr = np.mean(np.abs((y_test - y_pred_tr) / np.where(y_test == 0, 1, y_test))) * 100
    r2_tr = r2_score(y_test, y_pred_tr)

    # build comparison table
    df = pd.DataFrame({
        "model": ["ml", "traditional"],
        "mae": [mae_ml, mae_tr],
        "rmse": [rmse_ml, rmse_tr],
        "mape": [mape_ml, mape_tr],
        "r2": [r2_ml, r2_tr]
    })

    # save table
    df.to_pickle(output_table_path)

    print("saved:", output_table_path)
    print("comparison completed")
    print(df)

    return df
```
