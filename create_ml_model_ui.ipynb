```python
%%writefile /content/drive/MyDrive/ml_project/create_ml_model.py
# -*- coding: utf-8 -*-
import os
import ipywidgets as widgets
from IPython.display import display

target_dir = "/content/drive/MyDrive/ml_project"
choice_file = os.path.join(target_dir, "ml_method.txt")

# ---------------------------------------------------------
# create_ml_model_ui
# ---------------------------------------------------------
def create_ml_model_ui():

    # model selection list
    label_model = widgets.Label("select machine learning model:")
    dropdown_model = widgets.Dropdown(
        options=[
            "linear_regression",
            "ridge",
            "lasso",
            "random_forest_regressor",
            "svr",
            "xgboost_regressor",
            "mlp_regressor"
        ]
    )

    # parameter input
    label_param = widgets.Label("enter model parameters (comma separated if needed):")
    text_param = widgets.Text(
        placeholder="examples: ridge: alpha=1 ; random forest: n_estimators=100,max_depth=5"
    )

    # confirm button
    btn = widgets.Button(description="confirm")
    out = widgets.Output()

    def on_click(b):
        out.clear_output()

        model = dropdown_model.value
        params = text_param.value.strip()

        with open(choice_file, "w") as f:
            f.write(model + "\n")
            f.write(params + "\n")

        with out:
            print("saved:", choice_file)
            print("model:", model)
            print("parameters:", params)

    btn.on_click(on_click)

    display(
        widgets.VBox([
            label_model,
            dropdown_model,
            label_param,
            text_param,
            btn,
            out
        ])
    )
```
