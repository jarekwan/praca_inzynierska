```python
%%writefile /content/drive/MyDrive/ml_project/create_ml_model.py
# -*- coding: utf-8 -*-
import os
import ipywidgets as widgets
from IPython.display import display

target_dir = "/content/drive/MyDrive/ml_project"
choice_file = os.path.join(target_dir, "ml_method.txt")

# ---------------------------------------------------------
# create_ml_model_ui
# ---------------------------------------------------------
def create_ml_model_ui():

    # model selection
    label_model = widgets.Label("select machine learning model:")
    dropdown_model = widgets.Dropdown(
        options=[
            "linear_regression",
            "ridge",
            "lasso",
            "random_forest_regressor",
            "svr",
            "xgboost_regressor",
            "mlp_regressor"
        ]
    )

    # parameter mode selection
    label_mode = widgets.Label("select parameter selection mode:")
    dropdown_mode = widgets.Dropdown(
        options=[
            "manual",
            "grid_search",
            "random_search"
        ]
    )

    # manual parameter box
    label_manual = widgets.Label("enter manual parameters (comma separated):")
    text_manual = widgets.Text(
        placeholder="example: alpha=1,max_depth=5"
    )

    # grid search parameter box
    label_grid = widgets.Label("enter grid search parameter grid:")
    text_grid = widgets.Text(
        placeholder="example: alpha=[0.1,1,10]"
    )

    # random search parameter box
    label_random = widgets.Label("enter random search parameter ranges:")
    text_random = widgets.Text(
        placeholder="example: n_estimators=100-500, max_depth=3-12"
    )

    # confirm button
    btn = widgets.Button(description="confirm")
    out = widgets.Output()

    # hide all parameter fields initially
    label_manual.layout.display = "none"
    text_manual.layout.display = "none"
    label_grid.layout.display = "none"
    text_grid.layout.display = "none"
    label_random.layout.display = "none"
    text_random.layout.display = "none"

    # show only relevant parameter box based on mode
    def mode_change(change):
        mode = dropdown_mode.value

        # hide all fields
        label_manual.layout.display = "none"
        text_manual.layout.display = "none"
        label_grid.layout.display = "none"
        text_grid.layout.display = "none"
        label_random.layout.display = "none"
        text_random.layout.display = "none"

        # show selected field
        if mode == "manual":
            label_manual.layout.display = "block"
            text_manual.layout.display = "block"
        elif mode == "grid_search":
            label_grid.layout.display = "block"
            text_grid.layout.display = "block"
        elif mode == "random_search":
            label_random.layout.display = "block"
            text_random.layout.display = "block"

    dropdown_mode.observe(mode_change, names="value")

    def on_click(b):
        out.clear_output()

        model = dropdown_model.value
        mode = dropdown_mode.value

        # choose parameter string depending on mode
        if mode == "manual":
            params = text_manual.value.strip()
        elif mode == "grid_search":
            params = "grid:" + text_grid.value.strip()
        elif mode == "random_search":
            params = "random:" + text_random.value.strip()
        else:
            params = ""

        # save settings
        with open(choice_file, "w") as f:
            f.write(model + "\n")
            f.write(mode + "\n")
            f.write(params + "\n")

        with out:
            print("saved:", choice_file)
            print("model:", model)
            print("mode:", mode)
            print("parameters:", params)

    btn.on_click(on_click)

    display(
        widgets.VBox([
            label_model,
            dropdown_model,
            label_mode,
            dropdown_mode,
            label_manual,
            text_manual,
            label_grid,
            text_grid,
            label_random,
            text_random,
            btn,
            out
        ])
    )
```
