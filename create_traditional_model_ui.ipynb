```python
%%writefile /content/drive/MyDrive/ml_project/create_traditional_model.py
# -*- coding: utf-8 -*-
import os
import ipywidgets as widgets
from IPython.display import display

target_dir = "/content/drive/MyDrive/ml_project"
choice_file = os.path.join(target_dir, "traditional_method.txt")

# ---------------------------------------------------------
# create_traditional_model_ui
# ---------------------------------------------------------
def create_traditional_model_ui():

    # wybór modelu
    label_model = widgets.Label("select traditional forecasting model:")
    dropdown_model = widgets.Dropdown(
        options=["naive", "moving_average", "simple_exponential_smoothing", "arima"]
    )

    # pola parametrów
    # moving average
    label_ma = widgets.Label("moving average window:")
    text_ma = widgets.Text(placeholder="example: 3")

    # ses
    label_ses = widgets.Label("ses smoothing level (alpha) optional:")
    text_ses = widgets.Text(placeholder="example: 0.3")

    # arima
    label_arima = widgets.Label("arima parameters p,d,q:")
    text_arima = widgets.Text(placeholder="example: 1,1,1")

    # przycisk i pole wyjścia
    btn = widgets.Button(description="confirm")
    out = widgets.Output()

    # ukrycie wszystkich pól parametrów na start
    label_ma.layout.display = "none"
    text_ma.layout.display = "none"
    label_ses.layout.display = "none"
    text_ses.layout.display = "none"
    label_arima.layout.display = "none"
    text_arima.layout.display = "none"

    # zmiana widocznych pól zależnie od modelu
    def model_change(change):
        m = dropdown_model.value

        # reset
        label_ma.layout.display = "none"
        text_ma.layout.display = "none"
        label_ses.layout.display = "none"
        text_ses.layout.display = "none"
        label_arima.layout.display = "none"
        text_arima.layout.display = "none"

        # show fields
        if m == "moving_average":
            label_ma.layout.display = "block"
            text_ma.layout.display = "block"
        elif m == "simple_exponential_smoothing":
            label_ses.layout.display = "block"
            text_ses.layout.display = "block"
        elif m == "arima":
            label_arima.layout.display = "block"
            text_arima.layout.display = "block"

    dropdown_model.observe(model_change, names="value")

    # zapis wyboru
    def on_click(b):
        out.clear_output()

        m = dropdown_model.value

        if m == "naive":
            params = ""
        elif m == "moving_average":
            params = text_ma.value.strip()
        elif m == "simple_exponential_smoothing":
            params = text_ses.value.strip()
        elif m == "arima":
            params = text_arima.value.strip()
        else:
            params = ""

        with open(choice_file, "w") as f:
            f.write(m + "\n")
            f.write(params + "\n")

        with out:
            print("saved:", choice_file)
            print("model:", m)
            print("parameters:", params)

    btn.on_click(on_click)

    display(widgets.VBox([
        label_model,
        dropdown_model,
        label_ma,
        text_ma,
        label_ses,
        text_ses,
        label_arima,
        text_arima,
        btn,
        out
    ]))
```
