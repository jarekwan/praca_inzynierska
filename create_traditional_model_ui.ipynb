```python
%%writefile /content/drive/MyDrive/ml_project/create_traditional_model.py
# -*- coding: utf-8 -*-
import os
import ipywidgets as widgets
from IPython.display import display

target_dir = "/content/drive/MyDrive/ml_project"
choice_file = os.path.join(target_dir, "traditional_method.txt")

# ---------------------------------------------------------
# create_traditional_model_ui
# ---------------------------------------------------------
def create_traditional_model_ui():

    label_model = widgets.Label("select traditional forecasting model:")
    dropdown_model = widgets.Dropdown(
        options=["naive", "moving_average", "simple_exponential_smoothing", "arima"]
    )

    label_param = widgets.Label("enter model parameters (comma separated if needed):")
    text_param = widgets.Text(
        placeholder="example: moving average = 3 ; arima = 1,1,1"
    )

    btn = widgets.Button(description="confirm")
    out = widgets.Output()

    def on_click(b):
        out.clear_output()
        model = dropdown_model.value
        params = text_param.value.strip()

        with open(choice_file, "w") as f:
            f.write(model + "\n")
            f.write(params + "\n")

        with out:
            print("saved:", choice_file)
            print("model:", model)
            print("parameters:", params)

    btn.on_click(on_click)

    display(widgets.VBox([
        label_model,
        dropdown_model,
        label_param,
        text_param,
        btn,
        out
    ]))
```
