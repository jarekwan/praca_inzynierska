```python



%%writefile /content/drive/MyDrive/ml_project/date_features.py
# -*- coding: utf-8 -*-
import os
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

TARGET_DIR = "/content/drive/MyDrive/ml_project"
INPUT_PICKLE = os.path.join(TARGET_DIR, "df_clean.pkl")
CHOICE_FILE = os.path.join(TARGET_DIR, "date_features_col.txt")
OUTPUT_PICKLE = os.path.join(TARGET_DIR, "df_date_features.pkl")

# ---------------------------------------------------------
# date_features_ui
# wybór kolumny daty
# ---------------------------------------------------------
def date_features_ui():

    if not os.path.exists(INPUT_PICKLE):
        raise FileNotFoundError("df_clean.pkl not found")

    df = pd.read_pickle(INPUT_PICKLE)

    label_col = widgets.Label("select the date column for feature generation:")
    dropdown = widgets.Dropdown(options=df.columns.tolist())

    btn = widgets.Button(description="confirm")
    out = widgets.Output()

    def on_click(b):
        out.clear_output()
        date_col = dropdown.value

        with open(CHOICE_FILE, "w") as f:
            f.write(date_col + "\n")

        with out:
            print("saved:", CHOICE_FILE)
            print("selected date column:", date_col)

    btn.on_click(on_click)

    display(widgets.VBox([
        label_col,
        dropdown,
        btn,
        out
    ]))


# ---------------------------------------------------------
# date_features
# generuje: year, month, day, dayofweek
# ---------------------------------------------------------
def date_features():

    if not os.path.exists(INPUT_PICKLE):
        raise FileNotFoundError("df_clean.pkl not found")

    if not os.path.exists(CHOICE_FILE):
        raise FileNotFoundError("date_features_col.txt not found")

    df = pd.read_pickle(INPUT_PICKLE)

    with open(CHOICE_FILE, "r") as f:
        date_col = f.read().strip()

    df[date_col] = pd.to_datetime(df[date_col], errors="coerce")

    df["year"] = df[date_col].dt.year
    df["month"] = df[date_col].dt.month
    df["day"] = df[date_col].dt.day
    df["dayofweek"] = df[date_col].dt.dayofweek

    df.to_pickle(OUTPUT_PICKLE)
    print("saved:", OUTPUT_PICKLE)

    return df
```
