```python
%%writefile /content/drive/MyDrive/ml_project/date_features.py
# -*- coding: utf-8 -*-
import os
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

TARGET_DIR = "/content/drive/MyDrive/ml_project"
INPUT_PICKLE = os.path.join(TARGET_DIR, "df_clean.pkl")
CHOICE_FILE = os.path.join(TARGET_DIR, "date_features_col.txt")
OUTPUT_PICKLE = os.path.join(TARGET_DIR, "df_date_features.pkl")

# ---------------------------------------------------------
# date_features_ui
# user selects the date source for feature creation
# (column or index)
# ---------------------------------------------------------
def date_features_ui():

    if not os.path.exists(INPUT_PICKLE):
        raise FileNotFoundError("df_clean.pkl not found")

    df = pd.read_pickle(INPUT_PICKLE)

    options = df.columns.tolist()
    if df.index.name is not None:
        options = ["__index__"] + options

    label_col = widgets.Label(
        "select the date column (or index) for feature generation:"
    )
    dropdown = widgets.Dropdown(options=options)

    btn = widgets.Button(description="confirm")
    out = widgets.Output()

    def on_click(b):
        out.clear_output()
        date_col = dropdown.value

        with open(CHOICE_FILE, "w") as f:
            f.write(date_col + "\n")

        with out:
            print("saved:", CHOICE_FILE)
            print("selected date source:", date_col)

    btn.on_click(on_click)

    display(widgets.VBox([
        label_col,
        dropdown,
        btn,
        out
    ]))


# ---------------------------------------------------------
# date_features
# creates features: year, month, day, dayofweek
# ---------------------------------------------------------
def date_features():

    if not os.path.exists(INPUT_PICKLE):
        raise FileNotFoundError("df_clean.pkl not found")

    if not os.path.exists(CHOICE_FILE):
        raise FileNotFoundError("date_features_col.txt not found")

    df = pd.read_pickle(INPUT_PICKLE)

    with open(CHOICE_FILE, "r") as f:
        date_col = f.read().strip()

    if date_col == "__index__":
        date_series = pd.to_datetime(df.index, errors="coerce")
    else:
        date_series = pd.to_datetime(df[date_col], errors="coerce")

    df["year"] = date_series.year
    df["month"] = date_series.month
    df["day"] = date_series.day
    df["dayofweek"] = date_series.dayofweek

    df.to_pickle(OUTPUT_PICKLE)
    print("saved:", OUTPUT_PICKLE)

    return df
```
