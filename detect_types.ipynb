```python
# -*- coding: utf-8 -*-
import os
import pandas as pd
from google.colab import drive

# mount drive
drive.mount('/content/drive', force_remount=True)

TARGET_DIR = "/content/drive/MyDrive/ml_project"
FILTERED = os.path.join(TARGET_DIR, "filtered.pkl")
ORIGINAL = os.path.join(TARGET_DIR, "df.pkl")

def detect_types() -> dict[str, list[str]]:
    """
    loads the last saved dataframe from drive (filtered.pkl first, otherwise df.pkl)
    analyzes data types and returns groups of columns:
    numerical, categorical, datetime
    output: dict[str, list[str]]
    """

    # pick the correct dataframe
    if os.path.exists(FILTERED):
        data = pd.read_pickle(FILTERED)
    elif os.path.exists(ORIGINAL):
        data = pd.read_pickle(ORIGINAL)
    else:
        raise FileNotFoundError("no dataframe found on drive")

    # detect column groups
    numerical_cols = data.select_dtypes(include=["number"]).columns.tolist()
    categorical_cols = data.select_dtypes(include=["object", "category"]).columns.tolist()
    datetime_cols = data.select_dtypes(include=["datetime", "datetime64[ns]"]).columns.tolist()

    return {
        "numerical": numerical_cols,
        "categorical": categorical_cols,
        "datetime": datetime_cols
    }

# run detection
types = detect_types()
types
```
