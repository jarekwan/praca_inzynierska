```python
%%writefile /content/drive/MyDrive/ml_project/detect_types.py
# -*- coding: utf-8 -*-
import os
import pandas as pd

TARGET_DIR = "/content/drive/MyDrive/ml_project"
SELECTED_PICKLE = os.path.join(TARGET_DIR, "df_selected.pkl")
TYPES_OUTPUT = os.path.join(TARGET_DIR, "column_types.csv")


def detect_types() -> dict[str, list[str]]:
    """
    analyzes column types in df_selected.pkl
    returns: dict with keys: numerical, categorical, datetime
    saves result to column_types.csv
    """

    if not os.path.exists(SELECTED_PICKLE):
        raise FileNotFoundError("df_selected.pkl not found")

    df = pd.read_pickle(SELECTED_PICKLE)

    numerical = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
    datetime_cols = df.select_dtypes(include=["datetime64[ns]"]).columns.tolist()
    categorical = df.select_dtypes(include=["object", "category"]).columns.tolist()

    result = {
        "numerical": numerical,
        "categorical": categorical,
        "datetime": datetime_cols
    }

    # save results to csv for transparency
    rows = []
    for dtype, cols in result.items():
        for col in cols:
            rows.append([col, dtype])

    pd.DataFrame(rows, columns=["column", "type"]).to_csv(TYPES_OUTPUT, index=False)

    print("saved:", TYPES_OUTPUT)
    print(result)

    return result
```
