```python
%%writefile /content/drive/MyDrive/ml_project/evaluate_model.py
# -*- coding: utf-8 -*-
import numpy as np
import pandas as pd
import os

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

target_dir = "/content/drive/MyDrive/ml_project"
y_test_path = os.path.join(target_dir, "y_test.pkl")

metrics_output_path = os.path.join(target_dir, "metrics.txt")


# ---------------------------------------------------------
# evaluate_model
# ---------------------------------------------------------
def evaluate_model(y_pred: np.ndarray):

    if not os.path.exists(y_test_path):
        raise FileNotFoundError("y_test.pkl not found")

    # load test data
    y_test = pd.read_pickle(y_test_path)

    # convert to numpy
    y_test = np.array(y_test)

    # mae
    mae = mean_absolute_error(y_test, y_pred)

    # rmse
    rmse = mean_squared_error(y_test, y_pred, squared=False)

    # mape (bez dzielenia przez zero)
    mape = np.mean(np.abs((y_test - y_pred) / np.where(y_test == 0, 1, y_test))) * 100

    # r2
    r2 = r2_score(y_test, y_pred)

    metrics = {
        "mae": mae,
        "rmse": rmse,
        "mape": mape,
        "r2": r2
    }

    # save metrics
    with open(metrics_output_path, "w") as f:
        for k, v in metrics.items():
            f.write(f"{k}: {v}\n")

    print("saved:", metrics_output_path)
    print("evaluation completed")

    return metrics
```
