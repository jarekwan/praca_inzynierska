```python
%%writefile /content/drive/MyDrive/ml_project/load_data.py
# -*- coding: utf-8 -*-
import os
import pandas as pd
import requests

TARGET_DIR = "/content/drive/MyDrive/ml_project"

# ---------------------------------------------------------
# load_data
# loads data from upload.csv, upload.json or api_url.txt
# saves the result as df.pkl
# ---------------------------------------------------------
def load_data() -> pd.DataFrame:

    csv_path = os.path.join(TARGET_DIR, "upload.csv")
    json_path = os.path.join(TARGET_DIR, "upload.json")
    api_path = os.path.join(TARGET_DIR, "api_url.txt")

    # csv source
    if os.path.exists(csv_path):
        df = pd.read_csv(csv_path)

    # json source
    elif os.path.exists(json_path):
        df = pd.read_json(json_path)

    # api source
    elif os.path.exists(api_path):
        with open(api_path, "r") as f:
            url = f.read().strip()
        r = requests.get(url)
        r.raise_for_status()
        df = pd.DataFrame(r.json())

    else:
        raise FileNotFoundError("no saved data found in ml_project")

    # save dataframe as pickle
    df_pickle_path = os.path.join(TARGET_DIR, "df.pkl")
    df.to_pickle(df_pickle_path)

    print("DataFrame saved to:", df_pickle_path)
    return df
```
