```python
%%writefile /content/drive/MyDrive/ml_project/merge_predictions.py
# -*- coding: utf-8 -*-
import os
import pickle
import pandas as pd

target_dir = "/content/drive/MyDrive/ml_project"

# istniejące pliki
y_test_path = os.path.join(target_dir, "y_test.pkl")
y_trad_path = os.path.join(target_dir, "y_pred_traditional.pkl")
y_ml_path = os.path.join(target_dir, "y_pred_ml.pkl")

# wyjście
output_csv = os.path.join(target_dir, "final_predictions.csv")
output_pkl = os.path.join(target_dir, "final_predictions.pkl")


def merge_predictions():

    # wczytaj prawdziwe wartości
    y_true = pd.read_pickle(y_test_path).reset_index(drop=True)

    # wczytaj prognozy tradycyjne
    with open(y_trad_path, "rb") as f:
        y_pred_traditional = pickle.load(f)

    # wczytaj prognozy ML
    with open(y_ml_path, "rb") as f:
        y_pred_ml = pickle.load(f)

    # zbuduj tabelę
    df = pd.DataFrame({
        "y_true": y_true,
        "y_pred_traditional": y_pred_traditional,
        "y_pred_ml": y_pred_ml
    })

    # zapisz
    df.to_csv(output_csv, index=False)
    df.to_pickle(output_pkl)

    print("saved:", output_csv)
    print("saved:", output_pkl)

    return df
```
