```python
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from sklearn.neighbors import KNeighborsRegressor
import ipywidgets as widgets
from IPython.display import display
import pickle
from google.colab import drive

drive.mount('/content/drive')

def choose_model():
    model_dropdown = widgets.Dropdown(
        options=['SVR', 'Random Forest', 'K-Neighbors'],
        description='Model:',
    )

    select_button = widgets.Button(description="Select")
    output_area = widgets.Output()

    selected_model = None

    def on_button_click(b):
        nonlocal selected_model
        with output_area:
            output_area.clear_output()
            selected_algorithm = model_dropdown.value

            if selected_algorithm == 'SVR':
                selected_model = SVR()
            elif selected_algorithm == 'Random Forest':
                selected_model = RandomForestRegressor()
            elif selected_algorithm == 'K-Neighbors':
                selected_model = KNeighborsRegressor()
            else:
                raise ValueError("Invalid selection")

            print(f"Selected: {selected_algorithm} - model saved")

            # ZAPISZ model do Drive (WEWNĄTRZ with output_area)
            with open('/content/drive/MyDrive/ml_project/selected_model.pkl', 'wb') as f:
                pickle.dump(selected_model, f)
            print(f"Model saved to Google Drive")

    select_button.on_click(on_button_click)
    display(widgets.VBox([model_dropdown, select_button, output_area]))
```

```python
choose_model()
```
