```python
%%writefile /content/drive/MyDrive/ml_project/select_columns.py
# -*- coding: utf-8 -*-
import os
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

TARGET_DIR = "/content/drive/MyDrive/ml_project"
PICKLE_PATH = os.path.join(TARGET_DIR, "df.pkl")
OUTPUT_PATH = os.path.join(TARGET_DIR, "selected_columns.txt")

# ---------------------------------------------------------
# select_columns_ui
# UI with checkboxes for dataframe columns
# saves selected column names to selected_columns.txt
# ---------------------------------------------------------
def select_columns_ui(df: pd.DataFrame):
    label = widgets.Label("select columns for further analysis:")

    checkboxes = [
        widgets.Checkbox(value=False, description=col)
        for col in df.columns
    ]

    confirm_btn = widgets.Button(description="confirm")
    out = widgets.Output()

    def on_click(b):
        out.clear_output()

        selected = [cb.description for cb in checkboxes if cb.value]

        # save selected columns
        with open(OUTPUT_PATH, "w") as f:
            for col in selected:
                f.write(col + "\n")

        with out:
            print("saved to:", OUTPUT_PATH)
            print("selected columns:", selected)

    confirm_btn.on_click(on_click)

    display(widgets.VBox([label] + checkboxes + [confirm_btn, out]))


# ---------------------------------------------------------
# select_columns
# loads df.pkl and selected_columns.txt
# returns dataframe with chosen columns
# saves df_selected.pkl
# ---------------------------------------------------------
def select_columns() -> pd.DataFrame:

    if not os.path.exists(PICKLE_PATH):
        raise FileNotFoundError("df.pkl not found")

    if not os.path.exists(OUTPUT_PATH):
        raise FileNotFoundError("selected_columns.txt not found")

    df = pd.read_pickle(PICKLE_PATH)

    with open(OUTPUT_PATH, "r") as f:
        selected_cols = [x.strip() for x in f.readlines()]

    df_selected = df[selected_cols]

    out_path = os.path.join(TARGET_DIR, "df_selected.pkl")
    df_selected.to_pickle(out_path)

    print("saved:", out_path)
    print("selected columns:", selected_cols)

    return df_selected
```
