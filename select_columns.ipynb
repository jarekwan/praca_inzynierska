```python
# -*- coding: utf-8 -*-
import os
import pandas as pd
import ipywidgets as widgets
from IPython.display import display
from google.colab import drive

# mount drive
drive.mount('/content/drive', force_remount=True)

TARGET_DIR = "/content/drive/MyDrive/ml_project"
PICKLE_PATH = os.path.join(TARGET_DIR, "df.pkl")


# ---------------------------------------------------------
# select_columns_ui
# shows list of columns and returns selected names
# ---------------------------------------------------------
def select_columns_ui(data: pd.DataFrame) -> list[str]:
    """
    parameters: data
    what it does: displays the list of columns from the data table,
    asks the user which columns they want to choose for further analysis,
    and returns their names
    output: list[str] — names of the columns chosen by the user
    """

    instruction = widgets.Label("select columns for further analysis:")

    checkboxes = [
        widgets.Checkbox(value=False, description=col)
        for col in data.columns
    ]

    confirm_button = widgets.Button(description="confirm")
    output_box = widgets.Output()

    selected = []

    def on_confirm(b):
        output_box.clear_output()
        selected.clear()
        selected.extend([cb.description for cb in checkboxes if cb.value])
        with output_box:
            print("selected:", selected)

    confirm_button.on_click(on_confirm)

    display(widgets.VBox([instruction] + checkboxes + [confirm_button, output_box]))
    return selected

```

```python

# ---------------------------------------------------------
# select_columns
# returns dataframe with only selected columns
# ---------------------------------------------------------
def select_columns(data: pd.DataFrame, selected_cols: list[str]) -> pd.DataFrame:
    """
    parameters: data, selected_cols
    what it does: returns a dataframe containing only the selected columns
    output: pd.DataFrame
    """
    return data[selected_cols]


# ---------------------------------------------------------
# main execution: load df, run UI, filter, save result
# ---------------------------------------------------------

# load dataframe saved earlier
df = pd.read_pickle(PICKLE_PATH)

# run UI and get list of chosen column names
selected = select_columns_ui(df)

# build filtered dataframe
filtered_df = select_columns(df, selected)

# save filtered dataframe
FILTERED_PATH = os.path.join(TARGET_DIR, "filtered_df.pkl")
filtered_df.to_pickle(FILTERED_PATH)

print("saved:", FILTERED_PATH)
```
