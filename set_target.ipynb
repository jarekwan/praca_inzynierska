```python
%%writefile /content/drive/MyDrive/ml_project/set_target.py
# -*- coding: utf-8 -*-
import os
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

target_dir = "/content/drive/MyDrive/ml_project"
input_pickle = os.path.join(target_dir, "df_encoded.pkl")       # input from encode step
choice_file = os.path.join(target_dir, "target_method.txt")     # stores selected target
output_X = os.path.join(target_dir, "X.pkl")
output_y = os.path.join(target_dir, "y.pkl")

# ---------------------------------------------------------
# set_target_ui
# ---------------------------------------------------------
def set_target_ui():

    if not os.path.exists(input_pickle):
        raise FileNotFoundError("df_encoded.pkl not found")

    df = pd.read_pickle(input_pickle)

    label = widgets.Label("select the target column (y):")
    dropdown = widgets.Dropdown(options=df.columns.tolist())

    btn = widgets.Button(description="confirm")
    out = widgets.Output()

    def on_click(b):
        out.clear_output()

        col = dropdown.value

        with open(choice_file, "w") as f:
            f.write(col + "\n")

        with out:
            print("saved:", choice_file)
            print("target column:", col)

    btn.on_click(on_click)

    display(widgets.VBox([
        label,
        dropdown,
        btn,
        out
    ]))

# ---------------------------------------------------------
# set_target
# ---------------------------------------------------------
def set_target():

    if not os.path.exists(input_pickle):
        raise FileNotFoundError("df_encoded.pkl not found")

    if not os.path.exists(choice_file):
        raise FileNotFoundError("target_method.txt not found")

    df = pd.read_pickle(input_pickle)

    with open(choice_file, "r") as f:
        target_col = f.read().strip()

    if target_col not in df.columns:
        raise ValueError(f"target column '{target_col}' not found in dataframe")

    y = df[target_col]
    X = df.drop(columns=[target_col])

    X.to_pickle(output_X)
    y.to_pickle(output_y)

    print("saved:", output_X)
    print("saved:", output_y)
    print("X and y successfully created")

    return X, y
```
